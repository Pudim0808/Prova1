```python
# Perguntar a idade e o valor na conta
idade = int(input('Digite a sua idade: '))

# Lendo como texto e substituindo a vírgula por ponto para evitar erros
saldo_input = input('Digite o valor que tem na conta: ').replace(',', '.')
saldo = float(saldo_input)

# Verificar as condições
se_for_menor_de_16 = idade < 16

if se_for_menor_de_16:
    print('entrada apenas acompanhado')
else:
    if saldo >= 30:
        print('ingresso comprado')
    else:
        print('saldo insuficiente')

# Exibir mensagem final
print('foi bom te atender')
```
